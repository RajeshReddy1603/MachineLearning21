```python
import pandas as pd
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
df=pd.read_csv("pima-indians-diabetes.csv",header=None)
df
```

```python
x=df.iloc[:,0:8].values
y=df.iloc[:,8].values
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2)
X_train
```

```python
model=XGBClassifier()
model.fit(X_train,Y_train)
print(model.predict([[1,93,70,31,0,30.4,0.315,23]]))
```

```python
final=model.predict(X_test)
final
```

```python
accuracy=accuracy_score(Y_test,final)
accuracy*100
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_val_score
df=pd.read_csv("titanic.csv",usecols=["Pclass","Sex","Age","Fare","Survived"])
df
```

```python
x=df[["Age"]].mean()
df1=df.fillna(x).round(2)
df1
```

```python
dummies=pd.get_dummies(df1.Sex,dtype="int64")
dummies
```

```python
result=pd.concat([df1,dummies],axis="columns")
result
```

```python
result.drop(columns=["Sex","male"],axis=1,inplace=True)
result
```

```python
new_columns=["Pclass","Age","Fare","female","Survived"]
result=result[new_columns]
result
```

```python
x=result.drop("Survived",axis=1).values
x
```

```python
y=result["Survived"].values
y
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2)
```

```python
model=GaussianNB()
model.fit(X_train,Y_train)
print(model.predict(X_test))
```

```python
print(model.predict(X_test[0:5]))
```

```python
print(model.predict_proba(X_test[:5]))
```

```python
overall=cross_val_score(model,x,y,cv=10)
overall
```
